```python
import uproot
import numpy as np
import matplotlib.pyplot as plt
```

```python
# fp = '/storage/agrp/dreyet/FullEventReco/HGPflow_ATLAS/data/dijet_JZ1_to_JZ8_Jan2026/train_1M/train_jz123_bw-1.root'
fp = '/storage/agrp/dreyet/FullEventReco/HGPflow_ATLAS/data/dijet_JZ1_to_JZ8_Jan2026/train_1M/train_jzAll_bw-1.root'
tree = uproot.open(fp)['EventTree']
```

```python
n_particles = tree['nparticle'].array(library='np')
n_tracks = tree['ntrack'].array(library='np')
n_topos = tree['ntopo'].array(library='np')
```

```python
fig = plt.figure(figsize=(15,5))
gs = fig.add_gridspec(2,3)

ax1 = fig.add_subplot(gs[0, 0]); ax2 = fig.add_subplot(gs[1, 0])
for ax in [ax1, ax2]:
    bins = np.arange(0, np.max(n_particles)+2)-0.5
    ax.hist(n_particles, bins=bins, color='cornflowerblue', alpha=0.7)

ax3 = fig.add_subplot(gs[0, 1]); ax4 = fig.add_subplot(gs[1, 1])
for ax in [ax3, ax4]:
    bins = np.arange(0, np.max(n_tracks)+2)-0.5
    ax.hist(n_tracks, bins=bins, color='seagreen', alpha=0.7)

ax5 = fig.add_subplot(gs[0, 2]); ax6 = fig.add_subplot(gs[1, 2])
for ax in [ax5, ax6]:
    bins = np.arange(0, np.max(n_topos)+2)-0.5
    ax.hist(n_topos, bins=bins, color='salmon', alpha=0.7)

for ax in [ax2, ax4, ax6]:
    ax.set_yscale('log')
    ax.set_xlabel('Count')
    ax.set_ylabel('Number of Events')

ax1.set_title('#particles')
ax3.set_title('#tracks')
ax5.set_title('#topos')
```
